COLUMN-WISE DESCRIPTION:
1. REGION: The state/region of India where the data was collected.
2. DATE: The date/month for which the data was recorded.
3. FREQUENCY: How frequently the data is recorded(M-Monthly)
4. ESTIMATED UNEMPLOYMENT RATE: Percentage of the labour force that is unemployed and looking for work.
5. ESTIMATED EMPLOYED: Estimated number of people who are employed/working in that region during that period.
6. ESTIMATED LABOUR PARTICIPATION RATE: Percentage of the working-age population that is participating in the labour force (working or looking for work).
7. REGION: The geographical zone of India to which the state belongs.
8. LONGITUDE: Geographic longitude of the state/location used in the dataset.
9. LATITUDE: Geographic latitude of the state/location used in the dataset.

```python
sns.histplot(df["Estimated Unemployment Rate (%)"], kde=True)
plt.title("Distribution of Unemployment Rate")
plt.show()
```

The histogram shows that the unemployment rate is concentrated mainly between 0% and 20%, with most observations around 5–10%. The distribution is right-skewed, with a small number of observations having unusually high unemployment rates above 40%. These extreme observations may require further investigation as potential outliers.

```python
sns.boxplot(x=df["Estimated Unemployment Rate (%)"])
plt.title("Boxplot of Unemployment Rate")
plt.show()
```

From the boxplot, we can observe that the median unemployment rate lies approximately between 5% and 10%. There are also some observations with unemployment rates above 60%, which appear to be potential outliers.

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=df,
    x="Region",
    y="Estimated Unemployment Rate (%)",
    errorbar=None
)

plt.xticks(rotation=90)
plt.title("Unemployment Rate by Region")
plt.show()
```

From the above bar plot, we can get an idea of the average estimated unemployment rate across different regions. Haryana stands out with the highest average unemployment rate, at above 25%. Tripura ranks second, with an average unemployment rate of approximately 24%. Bihar and Jharkhand rank next, with average unemployment rates between 15% and 20%.

```python
corr = df.select_dtypes(include="number").corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x="Estimated Labour Participation Rate (%)",
    y="Estimated Unemployment Rate (%)"
)

plt.title("Labour Participation vs Unemployment Rate")
plt.show()
```

```python
plt.figure(figsize=(12, 6))

region_avg = df.groupby("Region")["Estimated Unemployment Rate (%)"].mean().reset_index()

sns.barplot(
    data=region_avg,
    x="Region",
    y="Estimated Unemployment Rate (%)",
    errorbar=None
)

plt.xticks(rotation=90)
plt.title("Average Unemployment Rate by Region")
plt.xlabel("Region")
plt.ylabel("Average Unemployment Rate (%)")
plt.show()
```

From the EDA, we observed that the unemployment rate varies across different regions. Haryana has the highest average unemployment rate, followed by Tripura. The unemployment rate also increased significantly during 2020, after which it gradually decreased.

```python

top_3 = (
    df.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
    .head(3)
    .index
)

print("Top 3 States:")
print(top_3)

# Filter only top 3 states
top_3_df = df[df["Region"].isin(top_3)].sort_values("Date")

# Line plot over time
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=top_3_df,
    x="Date",
    y="Estimated Unemployment Rate (%)",
    hue="Region",
    marker="o"
)

plt.title("Unemployment Rate Over Time - Top 3 States")
plt.xlabel("Date")
plt.ylabel("Estimated Unemployment Rate (%)")
plt.xticks(rotation=45)
plt.legend(title="State")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

# Calculate average unemployment rate for each region
region_avg = (
    df.groupby("Region")["Estimated Unemployment Rate (%)"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

# Bar plot
sns.barplot(
    data=region_avg,
    x="Region",
    y="Estimated Unemployment Rate (%)",
    errorbar=None
)

plt.xticks(rotation=45)
plt.title("Top 10 Regions by Average Unemployment Rate")
plt.xlabel("Region")
plt.ylabel("Average Unemployment Rate (%)")
plt.tight_layout()
plt.show()
```

```python

corr_data = df[
    [
        "Estimated Unemployment Rate (%)",
        "Estimated Employed",
        "Estimated Labour Participation Rate (%)"
    ]
]

correlation = corr_data.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Employment and Labour Market Indicators")
plt.show()
```

The heatmap shows that unemployment rate and labour participation rate have the strongest negative relationship, with a correlation of around -0.70. When the unemployment rate goes up, the labour participation rate tends to go down. Employment has only weak relationships with the other two variables.

```python

df["Date"] = pd.to_datetime(df["Date"])
pre_covid = df[df["Date"] < "2020-03-01"]
post_covid = df[df["Date"] >= "2020-03-01"]
pre_covid_mean = pre_covid["Estimated Unemployment Rate (%)"].mean()
post_covid_mean = post_covid["Estimated Unemployment Rate (%)"].mean()
print("Pre-COVID Mean Unemployment Rate:", pre_covid_mean)
print("Post-COVID Mean Unemployment Rate:", post_covid_mean)
```

### Pre-COVID vs Post-COVID Analysis
The mean unemployment rate before COVID-19 was **9.23%**, while the mean unemployment rate after COVID-19 increased to **12.96%**. This shows that the unemployment rate increased by approximately **3.73 percentage points** during the post-COVID period.
This indicates that the COVID-19 pandemic had a noticeable impact on employment conditions, leading to a higher unemployment rate after the onset of the pandemic.
